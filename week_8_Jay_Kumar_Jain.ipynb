# 🧠 Single Agent Pipeline Project

## Problem Statement
Build a **Single-Agent Smart Assistant** that:
- Understands user queries
- Routes tasks based on intent
- Uses tools when required
- Returns structured JSON output

### The agent should handle:
- Math queries → Calculator Tool
- Keyword extraction → Keyword Tool
- General queries → Direct response

---
### 🛠️ What You Need to Implement
- Agent logic
- Conditional routing
- Tool integration
- Basic error handling

### 🚀 Bonus
- Improve routing
- Add logging
- Add more tools


````python
# 🛠️ TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
````

````python
# 🛠️ TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()
        keywords = list(set([w.lower() for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []
````

````python
# 🎁 BONUS: Simple logging for the agent

agent_logs = []

def log_interaction(query: str, response: dict):
    """Store each query/response pair for later inspection."""
    agent_logs.append({"query": query, "response": response})
````

## 🤖 Implement Agent Logic Below

👉 Use conditional routing:
- If query contains "calculate" → use calculator
- If query contains "keywords" → use keyword extractor
- Else → general response

````python
# 🤖 AGENT FUNCTION (IMPLEMENTED)

import re

def agent(query: str):
    """
    Single-Agent Smart Assistant.

    Routes an incoming query to the correct tool based on intent:
      - "calculate" -> Calculator Tool
      - "keywords"  -> Keyword Extractor Tool
      - anything else -> General direct response

    Always returns a structured dict:
      {"type": "calculation" | "keywords" | "general" | "error", "result": ...}
    """
    try:
        if not isinstance(query, str) or not query.strip():
            response = {"type": "error", "result": "Empty or invalid query received."}
            log_interaction(query, response)
            return response

        query_lower = query.lower()

        # ---- Route 1: Calculator ----
        if "calculate" in query_lower:
            # Pull out the mathematical expression from the query
            match = re.search(r'[\d\.\+\-\*\/\(\)\s]+', query)
            expression = match.group().strip() if match else ""

            if not expression:
                response = {"type": "error", "result": "No valid expression found to calculate."}
            else:
                calc_result = calculator(expression)
                if calc_result == "Error in calculation":
                    response = {"type": "error", "result": f"Could not evaluate expression: '{expression}'"}
                else:
                    response = {"type": "calculation", "result": calc_result}

        # ---- Route 2: Keyword Extractor ----
        elif "keywords" in query_lower:
            # Remove the instruction phrase itself so it isn't extracted as a "keyword"
            text = re.sub(r'extract\s+keywords\s+from', '', query, flags=re.IGNORECASE).strip()

            if not text:
                response = {"type": "error", "result": "No text provided for keyword extraction."}
            else:
                keywords = extract_keywords(text)
                response = {"type": "keywords", "result": keywords}

        # ---- Route 3: General response ----
        else:
            response = {"type": "general", "result": f"I received your query: '{query}'. I don't have a specific tool for this, so here's a direct acknowledgment."}

    except Exception as e:
        response = {"type": "error", "result": f"Unexpected error: {str(e)}"}

    log_interaction(query, response)
    return response
````

```

  "result": ...
}
```
